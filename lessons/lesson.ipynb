```python
import cv2
import numpy as np
```

### Loading Images

```python
image = cv2.imread("../dataset/sports-car-clouds-digital-art-4k-wallpaper-uhdpaper.com-633@0@j.jpg")
cv2.imshow("Image", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Resizing an images

```python
new_width = 300
new_height = 200
resized_image = cv2.resize(image, (new_width, new_height))

cv2.imshow("Image", resized_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
scale_factor_x = 0.25  # Resizing to 50% of the original width
scale_factor_y = 0.25  # Resizing to 50% of the original height
resized_image = cv2.resize(image, (0, 0), fx=scale_factor_x, fy=scale_factor_y)

cv2.imshow("Image", resized_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Recoloring Images

```python
# Convert the image to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2.imshow('Grayscale Image', gray_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
center = (250, 250)  # (x, y) coordinates of the center
radius = 50  # Radius of the circle
color = (0, 255, 0)  # Color in BGR format (green)
thickness = 2  # Thickness of the circle border (-1 to fill the circle)

# Draw the circle on the image
cv2.circle(resized_image, center, radius, color, thickness)

# Display the image
cv2.imshow('Circle', resized_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Blurring Images

```python
image = cv2.imread("../dataset/sports-car-clouds-digital-art-4k-wallpaper-uhdpaper.com-633@0@j.jpg")

# Resizing the image
image = cv2.resize(image, (680, 420))

# Blurring the image
# args, image and kernel size
image = cv2.blur(image, (20, 20))

# Displaying images
cv2.imshow('image',image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Adding Filters to Images

```python
image = cv2.imread("../dataset/sports-car-clouds-digital-art-4k-wallpaper-uhdpaper.com-633@0@j.jpg")

# Resizing the image
image = cv2.resize(image, (680, 420))

# Making Kernel of dim 20x20
kernel = np.ones((20,20), np.float32) / 50

# Applying filter
image = cv2.filter2D(image, -1, kernel)

# Displaying images
cv2.imshow('image',image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Creating Sliders to visualize the effect of kernel and div on filter

```python
# Load and resize the image
image = cv2.imread("../dataset/sports-car-clouds-digital-art-4k-wallpaper-uhdpaper.com-633@0@j.jpg")
image = cv2.resize(image, (680, 420))

# Clone the original image to avoid overwriting
original_image = image.copy()

def filter_image(k, div):
    # Ensure the kernel size is at least 1 and an odd number
    k = max(1, k)
    if k % 2 == 0:
        k += 1
    
    # Avoid division by zero
    div = max(1, div)
    
    kernel = np.ones((k, k), np.float32) / div
    filtered_image = cv2.filter2D(original_image, -1, kernel)
    return filtered_image

# Create a window
cv2.namedWindow("Image")

# Create trackbars
cv2.createTrackbar('Kernel Size', 'Image', 1, 100, lambda x: None)
cv2.createTrackbar('Div', 'Image', 1, 50, lambda x: None)

while True:
    # Get current positions of trackbars
    k = cv2.getTrackbarPos('Kernel Size', 'Image')
    div = cv2.getTrackbarPos('Div', 'Image')
    
    # Apply filter
    filtered_image = filter_image(k, div)
    
    # Show the filtered image
    cv2.imshow('Image', filtered_image)
    
    # Break the loop when 'q' is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cv2.destroyAllWindows()
```
